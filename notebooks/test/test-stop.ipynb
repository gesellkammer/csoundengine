```python
import csoundengine as ce
```

```python
s = ce.Session()
```

```python
s.defInstr('sin', r'''
|kfreq=1000|
a0 = oscili(0.1, kfreq)
outch 1, a0
''')
```

```python
synth1 = s.sched('sin', whenfinished=lambda *args: print(f"finished!!, {args=}"))
synth1.stop(10)
synth1
```

```python
synth = s.sched('sin', kfreq=1005, delay=0.5)
```

```python
synth.stop(1)
synth
```
